```python
%%writefile sistem_panen.py

class KomoditasPertanian:
    def __init__(self, nama, harga):
        self.nama = nama
        self.__harga = harga

    def get_harga(self):
        return self.__harga

    def hitung_nilai(self, jumlah):
        return self.__harga * jumlah


class Beras(KomoditasPertanian):
    def __init__(self, nama, harga, kadar_air):
        super().__init__(nama, harga)
        self.kadar_air = kadar_air

    def hitung_nilai(self, jumlah):
        nilai = super().hitung_nilai(jumlah)

        if self.kadar_air > 14:
            nilai *= 0.9

        return nilai


beras = Beras("Beras Premium", 15000, 15)

print("Nama komoditas :", beras.nama)
print("Harga per kg   :", beras.get_harga())
print("Kadar air      :", beras.kadar_air, "%")
print("Nilai transaksi:", beras.hitung_nilai(10))
```
